### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top_5 = data.nlargest(5, 'IMDB_Rating')
top_5.set_index('Series_Title')[['IMDB_Rating', 'Meta_score']].plot(kind='bar', figsize=(10, 6))
plt.ylabel('Scores')
plt.title('Top 5 Movies: IMDB Rating (Scaled) vs Meta Score')
plt.xticks(rotation=45, ha='right')
plt.show()
```
